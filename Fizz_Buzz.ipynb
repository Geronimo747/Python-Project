Write a Python program to check if a number is positive, negative or zero

Write a Python program to check if a number entered by the user is odd or even.

Write a Python program to check if a number entered by the user is multiple of 3.

Write a Python program to check if a number entered by the user is multiple of 3 and 5.

```python
Number = int(input("Enter a number: "))
if Number > 0:
  print("Positive")
elif Number < 0:
  print("Negative")
else:
  print("Zero")

if Number % 2 == 0:
  print("Even")
else:
  print("Odd")

if Number % 3 == 0:
  print("Multiple of 3")

if Number % 3 == 0 and Number % 5 == 0:
  print("Multiple of 3 and 5")
```

Compares the number the user entered with a number you assigned

Provide a message “Little lower” or “Little higher” till the user guesses it correctly

```python
Number = 59
guess = int(input("Enter a number: "))
while guess != Number:
  if guess < Number:
    print("Little higher")
  else:
    print("Little lower")
  guess = int(input("Enter a number: "))
print("Correct")
```
